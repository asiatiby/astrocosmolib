```python
def hfunc_from_cosmology(z, cosmology):
    return cosmology.H(z).value
```

```python
import sys
sys.path.append("astrocosmolib/")
from pyACC.cosmology.distances import Distances
```

```python
mu_cmb = [0.01041, 0.22223 ,  0.14208]
cov_cmb = 1.e-9 * np.array([[0.006621, 0.12444, -1.1929] ,[0.12444, 21.344, -94.001], [ -1.1929, -94.001 , 1488.4 ] ])
inv_cov_cmb = np.linalg.inv(cov_cmb)
```

```python
def sound_horizon(omega_b, omega_bc, Neff):
  return 147.05*(omega_b/0.0236)**-0.13 * (omega_bc/0.1432)**-0.23 *(Neff/3.04)**-0.1

def model(params):
    rs= sound_horizon(params["omega_b"], params["omega_bc"], params["Neff"])
    cosmo_flat = FlatLambdaCDM(H0= params["H0"], Om0= params["omega_bc"])
    dist = Distances(lambda z: hfunc_from_cosmology(z, cosmo_flat))
    Dm = dist.comoving_distance(1089)
    return rs/Dm, params["omega_b"], params["omega_bc"]
```

```python
def cmb_log_likelihood(params):
  mu_pred= np.array(model(params))
  diff=  mu_pred - mu_cmb
  #prodotto scalare
  chi2= diff @ inv_cov_cmb @ diff
  return -0.5*chi2
```

```python
params={}
params["omega_b"]= 0.02236
params["omega_bc"]= 0.1432
params["Neff"]= 3.04
params["H0"]= 67.0
#params["H0"]= 70.0

model(params), cmb_log_likelihood(params)
```
