```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import trapezoid, simpson, quad
```

```python
def function(x):
  #return x**2 *np.sin(x) *np.exp(-x**2)
  return x**2 *np.sin(x)
```

```python
#xx= np.logspace(-4,3,100)
#fx = function(xx)
xx= np.linspace(0,100,100)
fx = function(xx)
```

```python
from re import X
plt.axhline(np.abs(integral_quad[0]), color= "r", linestyle="--")

list_n= np.logspace(2,4,20)
list_f=[]
list_s=[]
for n in list_n:
  xx= np.linspace(0,100,int(n))
  fx = function(xx)
  integral_trapz= trapezoid(fx,xx)
  integral_simps= simpson(fx,x=xx)

  list_f.append(integral_trapz)
  list_s.append(integral_simps)

plt.loglog(list_n, np.abs(list_f), label= "Trapezoid rule")
plt.loglog(list_n, np.abs(list_s), label="Simpson rule")
plt.legend(loc="lower right")
plt.xlabel("n")
plt.ylabel("f(x)")
plt.show()
```
